```python
# Alpha-Beta Pruning Algorithm

def alphabeta(depth, node, alpha, beta, maximizing):
    if depth == 3:
        return node

    if maximizing:
        best = -1000
        for value in [3, 5]:
            score = alphabeta(depth + 1, value,
                              alpha, beta, False)
            best = max(best, score)
            alpha = max(alpha, best)

            if beta <= alpha:
                break
        return best

    else:
        best = 1000
        for value in [3, 5]:
            score = alphabeta(depth + 1, value,
                              alpha, beta, True)
            best = min(best, score)
            beta = min(beta, best)

            if beta <= alpha:
                break
        return best


result = alphabeta(0, 0, -1000, 1000, True)
print("Optimal value:", result)

```
