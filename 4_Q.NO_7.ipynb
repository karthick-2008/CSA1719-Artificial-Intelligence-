```python
# Genetic Algorithm for Function Optimization

import random

# Objective function
def fitness(x):
    return x ** 2

# Generate initial population
population = [random.randint(0, 31) for _ in range(6)]

# Genetic Algorithm
for generation in range(10):

    # Calculate fitness
    population = sorted(population, key=fitness, reverse=True)

    # Selection
    parents = population[:3]

    # Create new population
    new_population = parents.copy()

    while len(new_population) < 6:
        p1, p2 = random.sample(parents, 2)

        # Crossover
        child = (p1 + p2) // 2

        # Mutation
        if random.random() < 0.2:
            child = random.randint(0, 31)

        new_population.append(child)

    population = new_population

    print("Generation", generation + 1,
          "Best Fitness:", fitness(max(population)))

# Final result
best = max(population, key=fitness)

print("\nBest Solution:", best)
print("Maximum Fitness:", fitness(best))
```
