```python
from itertools import product

# Define Bayesian Network probabilities
P_Rain = 0.2
P_Sprinkler = 0.3

# Conditional Probability Table
P_WetGrass = {
    (True, True): 0.99,
    (True, False): 0.90,
    (False, True): 0.80,
    (False, False): 0.01
}

# Calculate joint probability
def joint_probability(rain, sprinkler, wetgrass):
    p = P_Rain if rain else 1 - P_Rain
    p *= P_Sprinkler if sprinkler else 1 - P_Sprinkler

    pw = P_WetGrass[(rain, sprinkler)]
    p *= pw if wetgrass else 1 - pw

    return p

# Calculate probability of wet grass
total = 0
rain_and_wet = 0

for rain, sprinkler, wetgrass in product([True, False], repeat=3):
    p = joint_probability(rain, sprinkler, wetgrass)

    if wetgrass:
        total += p

    if rain and wetgrass:
        rain_and_wet += p

# Calculate conditional probability
result = rain_and_wet / total

print("Bayesian Network")
print("Variables: Rain, Sprinkler, WetGrass")
print("P(WetGrass) =", round(total, 4))
print("P(Rain and WetGrass) =", round(rain_and_wet, 4))
print("P(Rain | WetGrass) =", round(result, 4))
```
