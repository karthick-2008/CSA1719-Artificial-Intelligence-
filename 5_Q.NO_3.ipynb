```python
# Reinforcement Learning using Q-Learning

import numpy as np
import random

# Create environment
states = 5
actions = 2

# Initialize Q-table
Q = np.zeros((states, actions))

# Parameters
learning_rate = 0.8
discount_factor = 0.9
episodes = 100

# Training the agent
for episode in range(episodes):

    state = 0

    while state != 4:

        # Choose random action
        action = random.randint(0, 1)

        # Move to next state
        if action == 1:
            next_state = min(state + 1, 4)
        else:
            next_state = max(state - 1, 0)

        # Reward
        reward = 10 if next_state == 4 else -1

        # Update Q-table
        Q[state, action] += learning_rate * (
            reward + discount_factor * np.max(Q[next_state])
            - Q[state, action]
        )

        state = next_state

# Display Q-table
print("Trained Q-Table:")
print(np.round(Q, 2))

# Test the agent
state = 0
print("\nAgent Path:")

while state != 4:
    action = np.argmax(Q[state])

    if action == 1:
        next_state = min(state + 1, 4)
    else:
        next_state = max(state - 1, 0)

    print(state, "->", next_state)
    state = next_state

print("Goal Reached!")
```
