```python

import numpy as np

# XOR dataset
X = np.array([[0,0], [0,1], [1,0], [1,1]])
Y = np.array([[0], [1], [1], [0]])

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -50, 50)))

# Neural network prediction
def predict(ch):
    W1 = ch[:8].reshape(2, 4)
    b1 = ch[8:12]
    W2 = ch[12:16].reshape(4, 1)
    b2 = ch[16:]

    hidden = sigmoid(X @ W1 + b1)
    output = sigmoid(hidden @ W2 + b2)
    return output

# Fitness function
def fitness(ch):
    error = np.mean((Y - predict(ch)) ** 2)
    return 1 / (1 + error)

# Genetic Algorithm
np.random.seed(42)
population = np.random.uniform(-1, 1, (100, 17))

for generation in range(300):

    scores = np.array([fitness(ch) for ch in population])

    best = population[np.argmax(scores)].copy()

    if np.mean((Y - predict(best)) ** 2) < 0.01:
        break

    new_population = [best]

    while len(new_population) < 100:

        # Selection
        a = np.random.choice(100, 3, replace=False)
        p1 = population[a[np.argmax(scores[a])]]

        a = np.random.choice(100, 3, replace=False)
        p2 = population[a[np.argmax(scores[a])]]

        # Crossover
        mask = np.random.rand(17) < 0.5
        child = np.where(mask, p1, p2)

        # Mutation
        child += np.random.normal(0, 0.2, 17) * (
            np.random.rand(17) < 0.1
        )

        new_population.append(child)

    population = np.array(new_population)

# Best neural network
best = population[np.argmax(
    [fitness(ch) for ch in population]
)]

result = predict(best)

print("Generation:", generation)
print("Predicted Output:")
print(np.round(result, 3))

print("Final Classification:")
print((result >= 0.5).astype(int))
```
