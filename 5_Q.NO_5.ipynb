```python

import random

# Training sequence
sequence = ['A', 'B', 'A', 'C', 'B', 'A', 'B', 'C', 'A', 'B']

# Create transition model
transition = {}

for i in range(len(sequence) - 1):
    current = sequence[i]
    next_event = sequence[i + 1]

    if current not in transition:
        transition[current] = []

    transition[current].append(next_event)

# Predict next event
def predict_next(event):
    if event in transition:
        return random.choice(transition[event])
    else:
        return "No prediction available"

# Display transition model
print("Transition Model:")
for event, next_events in transition.items():
    print(event, "->", next_events)

# Predict next event
current_event = 'A'
prediction = predict_next(current_event)

print("\nCurrent Event:", current_event)
print("Predicted Next Event:", prediction)
```
