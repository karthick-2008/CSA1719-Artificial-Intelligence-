```python

import numpy as np
from sklearn.neural_network import MLPClassifier

# Input data
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 1])

# Create Feed Forward Neural Network
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='relu',
    solver='lbfgs',
    max_iter=1000,
    random_state=1
)

# Train the network
model.fit(X, y)

# Test the network
prediction = model.predict([[0, 1]])

print("Input:", [0, 1])
print("Predicted Output:", prediction[0])

```
